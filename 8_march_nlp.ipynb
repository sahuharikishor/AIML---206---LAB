## Representing  text as numerical data

```python
# example text as model training (sms messages)
simple_train = ['call you tonight', 'call me a cab', 'please call me', 'please call me...PLEASE']
```

```python
# Import and instantiate CountVectorizer (with the default parameter)
from sklearn.feature_extraction.text import CountVectorizer

vect = CountVectorizer()

# learn the 'vocabulary' of the training data (occurs in-place)
vect.fit(simple_train)

# examine the fitted vocabulary
vect.get_feature_names_out()
```

```python
# transform training data into a 'document-term matrix'
simple_train_dtm = vect.transform(simple_train)
simple_train_dtm
```

```python
# convert sparse matrix to a dense matrix
simple_train_dtm.toarray()
```

```python
# examine the vocabulary and document-term matrix together
pd.DataFrame(simple_train_dtm.toarray(), columns = vect.get_feature_names_out())
```

```python
# check the type of the document-term matrix
print(type(simple_train_dtm))

# examine the sparse matrix contents
print(simple_train_dtm)
```

```python
# example test for model testing
simple_test = ["please don't call me"]
```

```python
# transform testing data into a document-term matrix (using exesting vocablury)
simple_test_dtm = vect.transform(simple_test)
simple_test_dtm.toarray()
```

```python
# examine the vocabulary and document-term matrix together
pd.DataFrame(simple_test_dtm.toarray(), columns=vect.get_feature_names_out())
```

### Tokenization; Stop Word; Stemming; Lemitization

```python
#1} Tokenization : Tokens are formed
import nltk
nltk.download("punkt_tab")#punkt sentence tokenizer tokenizes the sentence and works on unsupervised learning and can be trained on unlabeled data.
#   Word tokenization
from nltk.tokenize import word_tokenize

text = "He stepped gingerly onto the bridge knowing, that enchantment awaited on the other side.\nIt was difficult for Mary to admit; that most of her workout\n consisted of exercising poor judgment"
tokens = word_tokenize(text)
print("\nThe sentence input used is : \n\n",text)
print("\n\nThe tokenized words are \n",tokens)
```

```python
# sentence tokenizer
from nltk.tokenize import sent_tokenize

tokens_sent = sent_tokenize(text)
print("\nThe tokenized sentence are: \n",tokens_sent)
```

```python
#2} Stop Words : Common words which are often removed during nlp usage

from nltk.corpus import stopwords
nltk.download("stopwords")

#nltk.download("stopwords")

stopwords= set(stopwords.words("english"))
print(stopwords)
filtered_words = []

for w in tokens:
    if w not in stopwords:
        filtered_words.append(w)
print("\nThe stop words are removed now, Tokenized words or filtered tokenized words are : \n\n",filtered_words)
```

```python
#3} Stemming : Reduces the words to their root form or base form
from nltk.stem import PorterStemmer
words = ["Organization","Organizational","Factual","running"]
#   list item stemming
stem_li = []
ps = PorterStemmer()
for i in words:
    stem_li.append(ps.stem(i))

print("The stemmed list items are: \t",stem_li)
```

```python
#   Stemming for words on word tokens
print(tokens)
stem_words = []
for i in tokens:
    stem_words.append(ps.stem(i))

print("\nThe stemmed list of tokenized words are : \t",stem_words)
```

```python
#   Stemming for senteces on sentenced tokens
print(tokens_sent)
stem_sentence = []
for i in tokens_sent:
    stem_sentence.append(ps.stem(i))

print("The stemmed list of tokenized sentence are : \t",stem_sentence)
```

```python
#4} Lemmatization : Breaks the words into simpler forms and makes meaning full words unlike stemming
from nltk.stem import WordNetLemmatizer
nltk.download('wordnet')
```

```python
wl=WordNetLemmatizer()

#   list item lemmitization
lemma_li=[]
for i in words:
    lemma_li.append(wl.lemmatize(i))

print("The lemmatized words of the list are : \t",lemma_li)
```

```python
lemma_words = []
for i in tokens:
    lemma_words.append(wl.lemmatize(i))

print("The lemmatized word of the sentence words tokens will be: \n",lemma_words)
```

```python
lemma_sentence = []
for i in tokens_sent:
    lemma_sentence.append(wl.lemmatize(i))

print("The lemmatized word of the sentence words tokens will be of the paragraph is : \n",lemma_sentence)
```

## Lemmatization

Stemming usually refers to a crude heuristic process that chops off the ends of words in the hope of achieving this goal correctly most of the time, and often includes the removal of derivational affixes.

Lemmatization usually refers to doing things properly with the use of a vocabulary and morphological analysis of words, normally aiming to remove inflectional endings only and to return the base or dictionary form of a word, which is known as the lemma. If confronted with the token saw, stemming might return just s, whereas lemmatization would attempt to return either see or saw depending on whether the use of the token was as a verb or a noun.

For example, runs, running, ran are all forms of the word run, therefore run is the lemma of all these words. Because lemmatization returns an actual word of the language, it is used where it is necessary to get valid words.

Python NLTK provides WordNet Lemmatizer that uses the WordNet Database to lookup lemmas of words.


```python
import nltk
nltk.download('wordnet')
```

```python
import nltk
from nltk.stem import WordNetLemmatizer

wordnet_lemmatizer = WordNetLemmatizer()

sentence = "He was running and eating at same time. He has bad habit of swimming after playing long hours in the Sun."
punctuations = "?:!.,;"

sentence_words = nltk.word_tokenize(sentence)

for word in sentence_words:
    if word in punctuations:
        sentence_words.remove(word)

sentence_words

print("{0:20}{1:20}".format("Word", "Lemma"))

for word in sentence_words:
    print("{0:20}{1:20}".format(word, wordnet_lemmatizer.lemmatize(word)))
```
