```python
def p_norm(vec,p):
    return np.sum(np.abs(vec)**p)**(1/p)
p_norm(v1,p)
```

```python
import numpy as np
v1=np.array([1,0])
v2=np.array([0,1])

# Angle between two vector
cos_theta=(v1@v2)/(p_norm(v1,2)*p_norm(v2,2))
angle=np.degrees(np.arccos(cos_theta))
print(f"The angle between two vectors {v1} and {v2} is {angle}")
```

```python
import numpy as np
v=np.array([1,2,3])

# Normalize the vector
normalization_v=v/p_norm(v,2)
print(f"The normalized vector is {normalization_v}")
print(np.linalg.norm(normalization_v,2))

```

## Basic operation of matrices

```python
import numpy as np
A=np.array([[2,1],[3,2]])
B= np.array([[1,1],[3,1]])

print(f"The Addition of two matrices \n {A} and \n {B} \n is {A+B}")
print(f"\nThe Elimentwise product of two matrices \n {A} and \n {B} \n is {A*B}")
print(f"\nThe Product  of two matrices \n {A} and \n {B} \n is {A@B}")
```

```python
import numpy as np
import sympy as sy
a,b,c,d,e,f,g,h,i=sy.symbols("a,b,c,d,e,f,g,h,i",real=True)
A_s=sy.Matrix([[a,b,c],[d,e,f],[g,h,i]])
B_s=sy.Matrix([[b,b,c],[d,e,f],[g,h,i]])
print(A_s*B_s)
A_s_new=sy.Matrix([[a,b],[d,e]])
print(f"The determinant of \n {A_s_new} is\n {A_s_new.det()}")
print(f"The Trace of \n {A_s_new} is\n {A_s_new.trace()}")
print(f"The Transpose of \n {A_s_new} is\n {A_s_new.transpose()}")
print(f"The Adjoint of \n {A_s_new} is\n {A_s_new.adjoint()}")
```

```python
# More Operations on matrices

A=np.array([[2,1],[3,2]])
print(A+5)
```

```python
# Transpose, Inverse, Determinant, Rank, Eigen Value and Eigen Vector 

A=np.array([[2,1],[3,2]])
print(f"Transpose of matrix\n {A} is\n {A.transpose()}")
print(f"Inverse of matrix\n {A} is\n {np.linalg.inv(A)}")
print(f"Determinant of matrix\n {A} is\n {np.linalg.det(A)}")
print(f"Rank of matrix\n {A} is\n {np.linalg.matrix_rank(A)}")
```

```python
# Eigen values and Eigen Vector

A= np.array([[2,1],[3,2]])
lam,v=np.linalg.eig(A)
print(f"The Eigen values of matrix\n {A} Is\n {lam}")
print(f"The Eigen values of matrix\n {A} Is\n {v}")
```
