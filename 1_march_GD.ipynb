## Gradient descent

```python
# Linear regression using gradient descent

import numpy as np
import matplotlib.pyplot  as plt

# Sample data
x=np.array([1,2,3,4,5])
y=np.array([3,5,7,9,11])

# Known slope
m=2

# Initial intercept
c = 0

# Parameters
eta = 0.01
epochs = 1000
n = len(x)

for i in range(epochs):

    # Gradient wrt c
    dc = (-2/n)*(np.sum(y-m*x)-n*c)

    # Update
    c = c - eta * dc

print("Optimized c:", c)
plt.scatter(x,y,label="Data")
plt.plot(x, m*x+c, color="red", label="Regression Line")
plt.legend()
plt.show()


```

```python
# Linear regression using gradient descent Optimized

import numpy as np
import matplotlib.pyplot  as plt

# Sample data
x=np.array([1,2,3,4,5])
y=np.array([3,5,7,9,11])


# Initialization of m and c
m = 0
c = 0

# Parameters
eta = 0.01
epochs = 1000  # we are getting close to 1
n = len(x)

for i in range(epochs):

    # slope wrt c
    dc = (-2/n)*(np.sum(y-m*x)-n*c)

    # slope wrt m
    dm = (-2/n) * np.sum(x*(y-m*x - c))


    # Update
    c = c - eta * dc
    m = m - eta * dm

print("Optimized c:", c)
print("Optimized m:", m)
plt.scatter(x,y,label="Data")
plt.plot(x, m*x+c, color="red", label="Regression Line")
plt.legend()
plt.show()
```
